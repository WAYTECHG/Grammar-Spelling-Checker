```python
spelling_pairs = []

with open(r"C:\vscode\AIT NLP grammar& spelling\Spelling.txt", "r", encoding="utf-8") as f:
    correct_word = None
    for line in f:
        word = line.strip()
        if not word:
            continue
        if word.startswith("$"):
            correct_word = word[1:].lower()
        else:
            wrong_word = word.lower()
            spelling_pairs.append((wrong_word, correct_word))
```

```python
import random

templates = [
    "I visited {} last year.",
    "They live in {}.",
    "He was born in {}.",
    "{} is a beautiful country.",
]

sentence_pairs = []
for wrong, correct in spelling_pairs:
    template = random.choice(templates)
    wrong_sentence = template.format(wrong)
    correct_sentence = template.format(correct)
    sentence_pairs.append((wrong_sentence, correct_sentence))
```

```python
from datasets import load_dataset

ds = load_dataset("jhu-clsp/jfleg")

grammar_pairs = []
for example in ds['test']: 
    original = example['sentence']
    for correction in example['corrections']:
        grammar_pairs.append((original.strip().lower(), correction.strip().lower()))
```

```python
all_pairs = sentence_pairs + grammar_pairs
```

```python
import csv

with open("combined_grammar_spelling_pairs.csv", "w", encoding="utf-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["input", "target"])  
    writer.writerows(all_pairs)

print("Saved to combined_grammar_spelling_pairs.csv")
```
